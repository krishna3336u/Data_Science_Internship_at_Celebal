**Your Geometric Explanation (4.3):**

Eigenvectors are special vectors that keep their direction when a matrix transformation is applied to them.  
The eigenvalue tells how much the vector is stretched or squished.

Geometrically:
- If the eigenvalue is greater than 1, the vector is stretched.
- If the eigenvalue is between 0 and 1, the vector is squished (shrunk).
- If the eigenvalue is negative, the vector also flips direction.

For the equation:

A v = λ v

the matrix A transforms the vector v into a new vector that points in the same direction as v, scaled by λ.

So, eigenvectors show the directions that remain unchanged by the transformation, while eigenvalues describe how much scaling happens in those directions.

**SVD → PCA Connection (4.4):**

In Singular Value Decomposition (SVD), a matrix A is decomposed as:

A=UΣV
T

where:

1. U matrix
Contains the left singular vectors.
Columns of U represent directions in the output space.
These vectors are orthogonal (perpendicular and unit length).
2. Σ (Sigma) matrix
A diagonal matrix containing the singular values.
Singular values tell how important each direction is.
Larger singular values correspond to more variance or information.
3. Vᵀ matrix
V
T
 is the transpose of V.
Columns of V are the right singular vectors.
These vectors represent the important directions in the original data space.
Relation to Principal Components (PCA)

In Principal Component Analysis (PCA):

The principal components are the directions where the data varies the most.
These directions come from the matrix:
V

(or rows of V
T
).

So:

V contains the directions of maximum variance.
Each column of V is a principal component direction.
The singular values in Σ indicate how much variance each principal component explains.
Geometric Meaning

SVD transforms data in three steps:

Rotate the data using V
T
Stretch/squish it using Σ
Rotate again using U

The directions of maximum variance are found in the columns of V because they capture the most important patterns in the data.

````python
from statsmodels.tsa.stattools import adfuller

np.random.seed(0)
s1 = np.random.normal(0, 1, 200)
s2 = np.random.exponential(1, 200)

# KS test — compare each against N(0,1)
ks_s1 = None  # YOUR CODE HERE
ks_s1 = stats.kstest(s1, 'norm')
ks_s2 = None  # YOUR CODE HERE
ks_s2 = stats.kstest(s2, 'norm')
print(f"KS test s1 (normal):      stat={ks_s1.statistic:.4f}, p={ks_s1.pvalue:.4f}")
print(f"KS test s2 (exponential): stat={ks_s2.statistic:.4f}, p={ks_s2.pvalue:.4f}")

# ADF test on non-stationary series
t = np.arange(200)
ts = 0.05 * t + np.random.normal(0, 1, 200)

adf_result = None  # YOUR CODE HERE — adfuller(ts)
adf_result = adfuller(ts)
print(f"\nADF statistic: {adf_result[0]:.4f}")
print(f"ADF p-value:   {adf_result[1]:.4f}")
print(f"Stationary?    {adf_result[1] < 0.05}")

# Difference once and re-run
ts_diff = None  # YOUR CODE HERE — np.diff
ts_diff = np.diff(ts)
adf_diff = None  # YOUR CODE HERE
adf_diff = adfuller(ts_diff)
print(f"\nAfter differencing — p-value: {adf_diff[1]:.4f}, Stationary? {adf_diff[1] < 0.05}")
````

````python
def compute_psi(expected, actual, bins=10):
    """
    PSI = sum((actual_pct - expected_pct) * ln(actual_pct / expected_pct))
    Add epsilon=1e-10 to avoid log(0).
    """
    # YOUR CODE HERE
    epsilon = 1e-10
    breakpoints = np.histogram_bin_edges(expected, bins=bins)
    expected_hist, _ = np.histogram(expected, bins=breakpoints, density=True)
    actual_hist, _ = np.histogram(actual, bins=breakpoints, density=True)
    expected_hist = np.clip(expected_hist, a_min=epsilon, a_max=None)
    actual_hist = np.clip(actual_hist, a_min=epsilon, a_max=None)
    expected_hist /= np.sum(expected_hist)
    actual_hist /= np.sum(actual_hist)


    psi = np.sum((actual_hist - expected_hist) * np.log((actual_hist + epsilon) / (expected_hist + epsilon)))

    return psi



np.random.seed(1)
train_dist = np.random.normal(50, 10, 1000)    # training distribution
drift_dist = np.random.normal(65, 12, 1000)    # shifted production distribution

psi_value = compute_psi(train_dist, drift_dist)
print(f"PSI: {psi_value:.4f}")
print(f"Shift severity: {'Major' if psi_value > 0.2 else 'Minor' if psi_value > 0.1 else 'Stable'}")

# Plot both distributions


fig, ax = plt.subplots()
# YOUR CODE HERE

ax.hist(train_dist, bins=20, alpha=0.5, label='Training Distribution')
ax.hist(drift_dist, bins=20, alpha=0.5, label='Production Distribution')
ax.legend(); ax.set_title('Distribution Shift Visualization')
plt.xlabel('Value'); plt.ylabel('Frequency')

plt.tight_layout(); plt.show()
````

**Concept Drift vs Covariate Drift & PSI Thresholds (5.5):**

Concept drift: When the relationship between input data and the target/output changes over time.

Example: Customer buying behavior changes after a festival season, so the old prediction model becomes less accurate.

Covariate drift: When the distribution of input features changes, but the relationship with the target stays mostly the same.

Example: The average age of customers changes, but spending behavior patterns remain similar.

PSI < 0.1: No significant drift; data is stable.

PSI 0.1–0.2: Moderate drift; monitor the model carefully.

PSI > 0.2: Significant drift; model retraining or investigation is needed.

Retraining trigger example:
If the PSI score for customer income changes from 0.08 to 0.25 and model accuracy drops, retrain the model using recent data.

### 6.4 Central Limit Theorem

1. Define a population following an **exponential distribution** (λ=1, size=100,000).
2. Repeatedly draw samples of size `n=30` and record their means — do this 5,000 times.
3. Plot the histogram of sample means.
4. Overlay the **theoretical normal distribution** predicted by the CLT: μ = population mean, σ = population_std / sqrt(n).
5. Run a KS test to confirm the distribution of sample means is approximately normal.

````python
np.random.seed(7)
population = np.random.exponential(scale=1.0, size=100_000)

n_samples = 5000
sample_size = 30
sample_means = None  # YOUR CODE HERE — array of 5000 sample means
sample_means = np.array([np.mean(np.random.choice(population, size=sample_size)) for _ in range(n_samples)])
pop_mean = population.mean()
pop_std  = population.std()
clt_std  = None  # YOUR CODE HERE — predicted std of sample means

clt_std = pop_std / np.sqrt(sample_size)

print(f"Population mean: {pop_mean:.4f}, Population std: {pop_std:.4f}")
print(f"Sample means mean: {sample_means.mean():.4f}, std: {sample_means.std():.4f}")
print(f"CLT predicted std: {clt_std:.4f}")

# Plot
fig, ax = plt.subplots()
# YOUR CODE HERE — histogram of sample means + theoretical normal overlay
ax.hist(sample_means, bins=20, density=True, alpha=0.5, label='Sample Means')
x = np.linspace(sample_means.min(), sample_means.max(), 100)
ax.plot(x, stats.norm.pdf(x, loc=pop_mean, scale=clt_std), label='CLT Theoretical')
ax.legend()
ax.set_xlabel('Sample Mean')
ax.set_ylabel('Frequency')

ax.set_title('CLT: Distribution of Sample Means')
plt.tight_layout(); plt.show()

# KS test
ks_result = stats.kstest(sample_means, 'norm', args=(pop_mean, clt_std))
print(f"KS test p-value: {ks_result.pvalue:.4f} → Approximately normal? {ks_result.pvalue > 0.05}")
````

::::::::**CLT Reflection (6.4):**

The Central Limit Theorem (CLT) is important because it states that the sampling distribution of the sample mean becomes approximately normal when the sample size is large, even if the original data is not normally distributed. This helps statisticians and machine learning practitioners make reliable predictions, estimate population parameters, and perform hypothesis testing.

The CLT justifies the assumption of normality used in many classical statistical tests such as z-tests, t-tests, confidence intervals, and linear regression analysis.
